### 1. Reformat Data
Load the master dataframe:

```python
df = pd.read_csv("outputs/master_dataframe.csv")
df
```

COMBAT prefers that our factors are one-hot encoded. We do the following:
* `sex` becomes `sex_male`, with TRUE = M and FALSE = F
* `chronotype` is split into `chrono_morning` (chronotype = 1), `chrono_evening` (chronotype = 2), with chronotype = 3 serving as the intercept.
* `protocol` becomes `total_deprivation`, with TRUE = SD protocol and FALSE = SR protocol.

```python
# SEX
df["sex_male"] = df.sex == "M"

# CHRONO
df["chrono_morning"] = df.chronotype == 1
df["chrono_evening"] = df.chronotype == 2

# PROTOCOL
df["total_deprivation"] = df.protocol == "SD"

df
```

We create individual dataframes from the different variable types:
* `Batch Feature`: the source of the variation we are trying to remove
* `Biological Features`: the features whose variation needs to be maintained
* `Anatomical/Diffusion Features`: the MRI-derived features that need to be corrected for the batch effect

Anatomical and diffusion covariates are harmonised separately because not all sites have DWI data.

```python
## ANAT
anat_subs = df["participant_id"]
anat_batch = df["site"]
anat_bio = df[["age","sex_male","kss_score_baseline","kss_score_sd","chrono_morning","chrono_evening"]]
anat_feat = df[["pvsvf_baseline","pvsvf_sd","cpvf_baseline","cpvf_sd"]]

## DWI
dwi_df = df.dropna(subset="alps_baseline")
dwi_subs = dwi_df["participant_id"]
dwi_batch = dwi_df["site"]
dwi_bio = dwi_df[["age","sex_male","kss_score_baseline","kss_score_sd","chrono_morning","chrono_evening"]]
dwi_feat = dwi_df[["mfwi_baseline","mfwi_sd","alps_baseline","alps_sd"]]
```

The model matrix cannot handle missing values, so we must check our biological features for NAs and impute missing values.

```python
for col in ["kss_score_baseline", "kss_score_sd"]:
    anat_bio.loc[:,col] = anat_bio[col].fillna(anat_bio[col].median())
    dwi_bio.loc[:,col] = dwi_bio[col].fillna(dwi_bio[col].median())
```

### 2. Harmonization
Create batch vectors:

```python
%%R -i anat_batch,dwi_batch
anat.batch <- as.numeric(as.factor(anat_batch))
dwi.batch <- as.numeric(as.factor(dwi_batch))
```

Create the model matrices:

```python
%%R -i anat_bio,dwi_bio
anat.mod <- model.matrix(~age+sex_male+kss_score_baseline+kss_score_sd+chrono_morning+chrono_evening, data=anat_bio)
dwi.mod <- model.matrix(~age+sex_male+kss_score_baseline+kss_score_sd+chrono_morning+chrono_evening, data=dwi_bio)
```

Check that each of the variables is the right size:

```python
%%R -i anat_feat
cat("nrow(anat.mod):", nrow(anat.mod), "\n")
cat("length(anat.batch):", length(anat.batch), "\n")
cat("ncol(t(anat_feat)):", ncol(t(anat_feat)), "\n")
```

Run harmonization for the anatomical features:

```python
%%R -i anat_feat,anat_subs
rownames(anat_feat) <- anat_subs
anat.harmonized <- neuroCombat(dat=t(anat_feat), batch=anat.batch, mod=anat.mod)
anat.harmonized
```

Run harmonization for the diffusion features:

```python
%%R -i dwi_feat,dwi_subs
rownames(dwi_feat) <- dwi_subs
dwi.harmonized <- neuroCombat(dat=t(dwi_feat), batch=dwi.batch, mod=dwi.mod)
dwi.harmonized
```

### 3. Output
Enter the adjusted values back into the original dataframe:

```python
%%R -i df -o df
# Initialize counter for changes
num_changes <- 0

for (sub in colnames(anat.adj)) { # Loop over the subjects...
  for (var in rownames(anat.adj)) { # ... and each of the anatomical features
    original_value <- df[df$participant_id == sub, var]
    adjusted_value <- anat.adj[var, sub]
    
    # Replace original values with adjusted values only if different
    if (!isTRUE(all.equal(original_value, adjusted_value))) {
      df[df$participant_id == sub, var] <- adjusted_value # Replace
      num_changes <- num_changes + 1
    }
  }
  
  if (sub %in% colnames(dwi.adj)) { # If this subject has dwi data
    for (var in rownames(dwi.adj)) { # Also loop over the diffusion features
      original_value <- df[df$participant_id == sub, var]
      adjusted_value <- dwi.adj[var, sub]
      
      # And replace those values only if different
      if (!isTRUE(all.equal(original_value, adjusted_value))) {
        df[df$participant_id == sub, var] <- adjusted_value
        num_changes <- num_changes + 1
      }
    }
  }
}

cat("Total values changed after harmonization:", num_changes, "\n")
```
