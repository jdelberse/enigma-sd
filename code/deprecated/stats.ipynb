### 1. Format Data
R prefers data in a narrow ("melted") format. This means we create two rows for each subject, one where `SD = 0` (i.e. the baseline condition) and one where `SD = 1` (i.e. the sleep deprived condition). This formatting will facilitate some later analyses in R.

```python
melted_data = harmonized_data.loc[harmonized_data.index.repeat(2)].reset_index(drop=True) # Duplicate rows
melted_data['sd'] = [0,1] * (int(len(melted_data)/2)) # Add SD condition

# KSS
melted_data['kss'] = melted_data['kss_score_baseline'] # Init KSS column
melted_data.loc[melted_data.sd == 1, "kss"] = np.nan # Set SD connection as NaN
melted_data['kss'].fillna(melted_data['kss_score_sd'], inplace=True)

# PVSVF
melted_data['pvsvf'] = melted_data["pvsvf_baseline"] # Init PVSVF column
melted_data.loc[melted_data.sd == 1, "pvsvf"] = np.nan # Set SD connection as NaN
melted_data['pvsvf'].fillna(melted_data["pvsvf_sd"], inplace=True)

# CPVF
melted_data['cpvf'] = melted_data["cpvf_baseline"] # Init CPVF column
melted_data.loc[melted_data.sd == 1, "cpvf"] = np.nan # Set SD connection as NaN
melted_data['cpvf'].fillna(melted_data["cpvf_sd"], inplace=True)

# RFWI
col_a = "mfwi_baseline"
col_b = "mfwi_sd"
melted_data.loc[(melted_data[col_a].isna()) ^ (melted_data[col_b].isna()), [col_a, col_b]] = np.nan
melted_data['mfwi'] = melted_data[col_a] # Init RWFI column
melted_data.loc[melted_data.sd == 1, "mfwi"] = np.nan # Set SD connection as NaN
melted_data['mfwi'].fillna(melted_data[col_b], inplace=True)

# DTI-ALPS
col_a = "alps_baseline"
col_b = "alps_sd"
melted_data.loc[(melted_data[col_a].isna()) ^ (melted_data[col_b].isna()), [col_a, col_b]] = np.nan
melted_data['alps'] = melted_data[col_a] # Init ALPS column
melted_data.loc[melted_data.sd == 1, "alps"] = np.nan # Set SD connection as NaN
melted_data['alps'].fillna(melted_data[col_b], inplace=True)

# Select new columns
melted_data = melted_data[["participant_id", "site", "group", "age", "sex_male", 
                           "chrono_morning", "chrono_evening", "sd", "kss", 
                           "pvsvf", "cpvf", "mfwi", "alps"]]
melted_data
```

Set the variables to the right data type:

```python
%%R -i melted_data
melted_data[c("site", "group", "sex")] <- lapply(melted_data[c("site", "group", "sex")], as.factor)
melted_data$sd <- as.logical(melted_data$sd)
melted_data
```
